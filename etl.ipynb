## Make Join Key Consistent Across all Files and Merge

Note that the NC Benfits DF Especially Needed Some Reworking. 

Originally we had something like:
| Plan  | Benefit          | Copay |
| ----- | ---------------- | ----- |
| 12345 | Specialist Visit | $50   |
| 12345 | Generic Drugs    | $10   |
 
After this cell we now have:
| Plan  | Specialist Visit | Generic Drugs |
| ----- | ---------------- | ------------- |
| 12345 | $50              | $10           |


```python
# Plan Attributes: keep just what is needed, one row per plan
plan_cols = ['StandardComponentId', 'IssuerId', 'IssuerMarketPlaceMarketingName',
             'PlanMarketingName', 'MetalLevel', 'PlanType',
             'TEHBDedInnTier1Individual', 'TEHBInnTier1IndividualMOOP', 'IsHSAEligible']
plans = nc_plan_attributes_df[plan_cols].drop_duplicates(subset='StandardComponentId')

# Rate: rename PlanId to match, keep core rating fields
rates = nc_rates_df.rename(columns={'PlanId': 'StandardComponentId'})
rate_cols = ['StandardComponentId', 'RatingAreaId', 'Tobacco', 'Age', 'IndividualRate']
rates = rates[rate_cols]

# BenCS: filter to a few benefits that matter, then pivot wide so that each row represents one insurance plan
key_benefits = ['Specialist Visit', 'Primary Care Visit', 'Generic Drugs', 'Emergency Room Services']
bencs_filtered = nc_benefits_df[nc_benefits_df['BenefitName'].isin(key_benefits)]
bencs_wide = bencs_filtered.pivot_table(
    index='StandardComponentId',
    columns='BenefitName',
    values='CopayInnTier1',
    aggfunc='first'
).reset_index()

# Merge everything
merged = plans.merge(rates, on='StandardComponentId', how='inner')
merged = merged.merge(bencs_wide, on='StandardComponentId', how='left')

print(merged.shape)
print(merged.head())
```

## Cleaning Money Columns So They Are No Longer Strings
Note that many plans use coinsurance (a %) instead of a flat copay for that benefit (mention in write up)

```python
import re

def parse_cost_sharing(val):
    """
    Parses insurance cost-sharing information and separates it into a 
    numerical value and a cost-sharing category.

    This function processes text-based cost-sharing values commonly found 
    in insurance plan datasets. It identifies dollar-based copays, 
    coinsurance percentages, no-charge services, missing values, and 
    unclear entries.

    Parameters
    val : str or object
        The raw cost-sharing value to parse. Values may include:
        - Dollar amounts (e.g., "$50", "$1,000 copay")
        - Coinsurance percentages (e.g., "20% coinsurance")
        - Missing values (e.g., NaN, None)
        - "Not Applicable"
        - "No Charge"

    Returns
    tuple
        A tuple containing:
        
        - Parsed numerical cost value (float or pd.NA)
            - Returns the dollar amount for copays.
            - Returns 0.0 for no-charge services.
            - Returns pd.NA when no numerical dollar value is available.

        - Cost-sharing type (str)
            - "Copay" for dollar-based cost-sharing.
            - "No Charge" for free services.
            - "[percentage]% Coinsurance" for percentage-based costs.
            - "Missing" for unavailable values.
            - "Not Applicable" when the service does not apply.
            - "Other/Unclear" when the value cannot be categorized.

    Examples
    >>> parse_cost_sharing("$50 copay")
    (50.0, "Copay")

    >>> parse_cost_sharing("20% coinsurance")
    (pd.NA, "20% Coinsurance")

    >>> parse_cost_sharing("No Charge")
    (0.0, "No Charge")

    >>> parse_cost_sharing("Not Applicable")
    (pd.NA, "Not Applicable")

    Notes
    Regular expressions are used to extract dollar amounts and percentage
    values from text strings. Dollar amounts are converted to floats and
    commas are removed for numerical processing.
    """

    val = str(val).strip()
    
    if val in ('nan', 'None'):
        return pd.NA, 'Missing'
    if val == 'Not Applicable':
        return pd.NA, 'Not Applicable'
    if 'no charge' in val.lower():
        return 0.0, 'No Charge'
    
    dollar_match = re.search(r'\$([\d,]+\.?\d*)', val)
    if dollar_match:
        amount = float(dollar_match.group(1).replace(',', ''))
        return amount, 'Copay'
    
    pct_match = re.search(r'(\d+)%', val)
    if pct_match:
        return pd.NA, f'{pct_match.group(1)}% Coinsurance'
    
    return pd.NA, 'Other/Unclear'

money_cols = ['TEHBDedInnTier1Individual', 'TEHBInnTier1IndividualMOOP',
              'Emergency Room Services', 'Generic Drugs', 'Specialist Visit']

for col in money_cols:
    parsed = merged[col].apply(parse_cost_sharing)
    merged[col] = parsed.apply(lambda x: x[0])
    merged[f'{col}_Type'] = parsed.apply(lambda x: x[1])
```
